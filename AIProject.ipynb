```python
import google.generativeai as genai
from google.colab import userdata
import json
import time

genai.configure(api_key=userdata.get('GEMINI_API_KEY'))
print("Gemini initialized successfully.")
```

# Lab 1

```python
def get_completion(prompt, system_instruction="You are a helpful assistant."):
    m = genai.GenerativeModel(
        model_name="gemini-2.5-flash",
        system_instruction=system_instruction
    )
    response = m.generate_content(prompt)
    time.sleep(12)  # stays under free tier rate limit (5 requests/min)
    return response.text

prompt = input("Enter a prompt: ")
response = get_completion(prompt)
print("\n---Response ---")
print(response)
```

```python
def get_completion(prompt, system_instruction="You are a helpful assistant."):
    m = genai.GenerativeModel(
        model_name="gemini-2.5-flash",
        system_instruction=system_instruction
    )
    response = m.generate_content(prompt)
    time.sleep(12)  # stays under free tier rate limit (5 requests/min)
    return response.text

# A realistic prompt
system_prompt = """Design an system to address poverty in the Bay Area for users. Define: input (financial + resource data), AI processing (analyze needs/barriers),
output (personalized resources/recommendations),
and real-world action (improving housing, income, or education access)."""

issue_complaint = (
    "Describe Jennifer, an African American receptionist at San Jose City Hall,"
    "focusing on her lived experience with poverty in the Bay Area. Explain her financial challenges,"
    "the systemic barriers she faces (housing costs, education, income), and how these factors limit her opportunities and economic mobility."
    )

response = get_completion(issue_complaint, system_prompt)

print("--- Issue complaint ---")
print(issue_complaint)
print("\n--- Response ---")
print(response)
```

```python
resident_message = ("As an African American woman, Jennifer faces an additional layer of systemic challenges. Historical discriminatory housing policies (redlining) have limited wealth accumulation in Black communities, making it harder to pass down assets or even simply secure housing in desirable areas. Implicit biases in hiring and promotion can also create subtle but persistent barriers to career advancement, even in public sector jobs. She often feels invisible amidst the Bay Area's overwhelmingly white and Asian tech workforce, a stark reminder of the region's economic segregation.")

def extract_unstructured(message):
    m = genai.GenerativeModel(
        model_name="gemini-2.5-flash",
        system_instruction="Come up with 3 ways to tackle the issue and dismantle the racial disparities"
    )
    response = m.generate_content(message)
    time.sleep(12)  # stays under free tier rate limit
    return response.text

print("--- Issue At Hand ---")
print(resident_message)
print("\n--- Structured Extraction ---")
print("\nOutput:")
print(extract_unstructured(resident_message))
```
